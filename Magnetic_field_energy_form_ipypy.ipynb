```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
print("MAGNETIC FIELD ENERGY MODEL")
print("---------------------------")

L1 = float(input("Enter L1: "))
L2 = float(input("Enter L2: "))
M = float(input("Enter mutual inductance M: "))

I1 = float(input("Enter current I1: "))
I2 = float(input("Enter current I2: "))
```

```python
L = np.array([
    [L1, M],
    [M, L2]
])

I = np.array([I1, I2])

print("\nInductance Matrix L:")
print(L)

print("\nCurrent Vector I:")
print(I)
```

```python
energy = 0.5 * I.T @ L @ I

print("\nMagnetic Field Energy:")
print(round(float(energy), 4))
```

```python
eigenvalues = np.linalg.eigvalsh(L)

print("\nEigenvalues:")
for value in eigenvalues:
    print(round(float(value), 4))
```

```python
if np.all(eigenvalues > 0):
    print("\nThe matrix is POSITIVE DEFINITE.")
else:
    print("\nThe matrix is NOT POSITIVE DEFINITE.")
```

```python
det_L = np.linalg.det(L)

print("\nMATHEMATICAL CHECK")
print("------------------")

print("L1 × L2 - M²")
print("=", L1 * L2, "-", M**2)

print("=", round(float(det_L), 4))

print("\nEigenvalues:")
print(np.round(eigenvalues, 4))

print("\nConclusion:")

if L1 > 0 and det_L > 0:
    print("L1 > 0 and determinant > 0")
    print("Therefore, L is POSITIVE DEFINITE.")
else:
    print("The matrix is NOT positive definite.")
```

```python
i1_values = np.linspace(-3, 3, 50)
i2_values = np.linspace(-3, 3, 50)

I1_grid, I2_grid = np.meshgrid(i1_values, i2_values)

Energy_grid = 0.5 * (
    L1 * I1_grid**2
    + 2 * M * I1_grid * I2_grid
    + L2 * I2_grid**2
)

plt.figure(figsize=(7, 5))

plt.contourf(
    I1_grid,
    I2_grid,
    Energy_grid,
    levels=20
)

plt.colorbar(label="Magnetic Energy")

plt.xlabel("Current I1")
plt.ylabel("Current I2")
plt.title("Magnetic Field Energy")

plt.show()
```
